```python
import pandas as pd
import numpy as np

data_transaksi = {
    'ID_Transaksi': ['TRX01', 'TRX02', 'TRX03', 'TRX04', 'TRX05', 'TRX06', 'TRX07', 'TRX08', 'TRX09', 'TRX10'],
    'Cabang': ['Depok', 'Jakarta', 'Bandung', 'Depok', 'Jakarta', 'Depok', 'Bandung', 'Jakarta', 'Depok', 'Bandung'],
    'Produk': ['Laptop', 'Mouse', 'Monitor', 'Keyboard', 'Laptop', 'Monitor', 'Headset', 'Keyboard', 'Mouse', 'Laptop'],
    'Kategori': ['Komputer', 'Aksesoris', 'Komputer', 'Aksesoris', 'Komputer', 'Komputer', 'Aksesoris', 'Aksesoris', 'Aksesoris', 'Komputer'],
    'Harga_Satuan': [12000000, 250000, 3000000, 500000, 12500000, np.nan, 400000, 550000, 250000, 11800000],
    'Qty': [2, 10, -3, 5, 1, 4, 8, -2, 15, 2],
    'Diskon_Persen': [0.10, 0.0, 0.05, 0.0, 0.10, 0.05, 0.05, 0.0, 0.05, 0.15]
}

df = pd.DataFrame(data_transaksi)
df
```

# Indexing & Selection




```python
# Memperbaiki nilai Qty negatif menjadi posisitif
df.loc[df['Qty'] < 0, 'Qty'] = df.loc[df['Qty'] < 0, 'Qty'].abs()
df
```

```python
# Meperbaiki nilai NaN
df.loc[df['Harga_Satuan'].isna(), 'Harga_Satuan'] = 3000000
df
```

```python
# Mengambil 3 baris pertama dan kolom ID_Transaksi, Produk, Qty
subset_iloc = df.iloc[:3, [0, 2, 5]]
subset_iloc
```

# Operasi Aritmatika

```python
# Menghitung Total_Kotor
df['Total_Kotor'] = df['Harga_Satuan'] * df['Qty']
df
```

```python
# Menghitung Nominal_Diskon
df['Nominal_Diskon'] = df['Total_Kotor'] * df['Diskon_Persen']
df
```

```python
# Menghitung Total_Bersih
df['Total_Bersih'] = df['Total_Kotor'] - df['Nominal_Diskon']
df
```

# Filtering

```python
# Filter cabang Depok/Bandung yang Total_Bersih diatas 5jt
filter_cabang_besar = df[
    (df['Cabang'].isin(['Depok', 'Bandung']))&
    (df['Total_Bersih'] > 5000000)]
filter_cabang_besar
```

```python
# Filter menggunakan query() untuk kategori Aksesoris dan Qty >= 8
filter_aksesoris = df.query('Kategori == "Aksesoris" and Qty >= 8')
filter_aksesoris
```

# Sorting

```python
# Multi-column sorting
df_sorted = df.sort_values(by=['Cabang', 'Total_Bersih'], ascending=[True, False])
df_sorted
```

```python
# Mengambil Top 3 transaksi berdasarkan Total_Bersih
top_3_transaksi = df.nlargest(3, 'Total_Bersih')
top_3_transaksi
```

# Grouping & Agregasi

```python
# Named Aggregation per Cabang, diurutkan dari Total_Omzet_Bersih Tertinggi
laporan_cabang = df.groupby('Cabang', as_index=False).agg(
    Jumlah_Transaksi=('ID_Transaksi', 'count'),
    Total_Unit_Terjual=('Qty', 'sum'),
    Total_Omzet_Bersih=('Total_Bersih', 'sum'),
    Rata_Rata_Transaksi=('Total_Bersih', 'mean')
).sort_values(by='Total_Omzet_Bersih', ascending=False)
laporan_cabang
```

```python
# Menghitung presentasi kontribusi Transaksi terhadap total omzet cabangnya
total_per_cabang = df.groupby('Cabang')['Total_Bersih'].transform('sum')
df['Kontribusi_Cabang_Persen'] = df['Total_Bersih'] / total_per_cabang * 100
df
```
